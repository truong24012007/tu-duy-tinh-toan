W6A16:Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input().split())

a = []

for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

W6A17:Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())
matran = []
for i in range(n):
    hang = list(map(int, input().split()))
    matran.append(hang)
for i in range(n):
    for j in range(n):
        print(matran[i][j], end = ' ')
    print('')
print('Đường chéo chính:')    
for i in range(n):
    print(matran[i][i], end = ' ')
print()    
print('Đường chéo phụ:') 
for i in range(n):
    print(matran[i][n - 1 - i], end = ' ')  
```

W6A18: Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())
matran = []
for i in range(n):
    hang = list(map(int, input().split()))
    matran.append(hang)
for i in range(n):
    for j in range(m):
        print(matran[i][j], end = ' ')
    print('')
tong_cot = 0
for x in range(n):
    tong_cot += matran[x][k]  
print('Tổng cột', k, '=', tong_cot)   
tong_hang = 0
for y in range(m):
    tong_hang += matran[k][y] 
print('Tổng hàng', k, '=', tong_hang)
```

W6A19:Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất

W6A20:Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def vitri(lst, k):
  for i in range(len(lst)):
     if lst[i] == k:
        return i
  return -1
lst = list(map(int, input().split()))
k = int(input())  
print(vitri(lst, k))
```

W6A21:Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
lst = input().split()
d = {}
for i in range(len(lst)):
    if lst[i] not in d:
        d[lst[i]] = (i,)
    else:
        d[lst[i]] += (i,) 
print(d)        
```

W6A23:VIết hàm xóa đi các số <x có trong mảng

```python
def xoa(lst, x):
    lst1 = []
    for i in range(len(lst)):
        if lst[i] >= x:
            lst1.append(lst[i])
    return lst1
lst = list(map(int, input().split()))
x = int(input())
print(xoa(lst, x))
```

W6A24:Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9


```python
lst = list(map(int, input().split()))
lst1 = []
lst2 = []
for x in lst:
    if x == 0:
        lst1.append(x)
    else:
        lst2.append(x)    
lst3 = lst1 + lst2
lst4 = lst2 + lst1
print(lst3)
print(lst4)        
```

W6A26:Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
lst = list(map(int, input().split()))
tong = 0
for i in range(1, len(lst) - 1):
    if lst[i] > lst[i - 1] and lst[i] > lst[i + 1]:
        tong += lst[i]
print(tong)        
```

W6A27:Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1

```python
lst = list(map(int, input().split()))
x = int(input())
found = False
for i in range(len(lst)):
    if lst[i] < x:
        print(i, end = ' ')
        found = True
if not found:
    print('-1')        
```

W6A28:Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
lst = list(map(int, input().split()))

tang = True
giam = True

for i in range(len(lst) - 1):
    if lst[i] >= lst[i + 1]:
        tang = False
    if lst[i] <= lst[i + 1]:
        giam = False

if tang:
    print("Tang thuc su")
elif giam:
    print("Giam thuc su")
else:
    print("Vo danh")
```
