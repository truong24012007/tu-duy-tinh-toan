sách được sắp xếp tăng dần.
Quy tắc:
- Nếu a == 0 và b != 0: trả về nghiệm duy nhất [-c / b].
- Nếu a == 0 và b == 0: trả về [] (không có nghiệm hữu hạn duy nhất nếu c == 0, hoặc vô nghiệm nếu c != 0).
- Với a != 0, dùng biệt thức d = bb - 4ac:
- d < 0 → []
- d == 0 → [ -b / (2a) ]
- d > 0 → hai nghiệm [r1, r2] với r1 <= r2

```python
import math

def giai_pt_bac_2(a, b, c):
    if a == 0:
        if b != 0:
            return [-c / b]
        else:
            return [] 
    d = b * b - 4 * a * c

    if d < 0:
        return []
    elif d == 0:
        return [-b / (2 * a)]
    else:
        sqrt_d = math.sqrt(d)
        r1 = (-b - sqrt_d) / (2 * a)
        r2 = (-b + sqrt_d) / (2 * a)
        return sorted([r1, r2])
```

```python
import math

def C(n, k):
    if k < 0 or k > n:
        return 0
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
```

Bài 16:Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3). 


```python
import math

def goc_ABC(x1, y1, x2, y2, x3, y3):
    # vector BA và BC
    BAx, BAy = x1 - x2, y1 - y2
    BCx, BCy = x3 - x2, y3 - y2
    
    # tích vô hướng
    dot = BAx * BCx + BAy * BCy
    
    # độ dài
    len_BA = math.sqrt(BAx**2 + BAy**2)
    len_BC = math.sqrt(BCx**2 + BCy**2)
    
    # tránh chia cho 0
    if len_BA == 0 or len_BC == 0:
        return None
    
    # cos góc
    cos_theta = dot / (len_BA * len_BC)
    
    # tránh sai số làm cos > 1 hoặc < -1
    cos_theta = max(-1, min(1, cos_theta))
    
    # đổi sang độ
    theta = math.degrees(math.acos(cos_theta))
    
    return theta
```

Bài 25: Yêu cầu: Viết hàm point_in_quadrilateral(quad, p) trả về True nếu điểm p nằm bên trong hoặc trên biên của tứ giác quad,
và False nếu ngược lại.
- Đầu vào:
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm,
nhưng không tự cắt).
- p: một điểm (x, y).
- Đầu ra: giá trị boolean.
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm
trên biên là ở trong. 


```python
def point_in_quadrilateral(quad, p):
    x, y = p
    n = 4
    inside = False

    def on_segment(x1, y1, x2, y2, x, y):
        return (
            min(x1, x2) <= x <= max(x1, x2) and
            min(y1, y2) <= y <= max(y1, y2) and
            (x2 - x1) * (y - y1) == (y2 - y1) * (x - x1)
        )

    for i in range(n):
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % n]

        # kiểm tra nằm trên cạnh
        if on_segment(x1, y1, x2, y2, x, y):
            return True

        # ray casting
        if ((y1 > y) != (y2 > y)):
            xinters = x1 + (y - y1) * (x2 - x1) / (y2 - y1)
            if xinters >= x:
                inside = not inside

    return inside
```

Bài 26:Yêu cầu: Viết hàm are_collinear(x1, y1, x2, y2, x3, y3) trả về True nếu ba điểm thẳng hàng, ngược lại trả về False.
- Đầu vào: ba điểm x1, y1, x2, y2, x3, y3 biểu diễn tọa độ của ba điểm.
- Đầu ra: giá trị boolean.
- Gợi ý: Dùng phép kiểm tra diện tích (tích có hướng) và một sai số nhỏ khi so sánh số thực. Các điểm nằm trên cùng một
đường thẳng được xem là thẳng hàng (kể cả khi có điểm trùng nhau). 


```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    s = x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)
    
    # sai số nhỏ cho số thực
    eps = 1e-9
    
    return abs(s) < eps
```

- Đầu vào: các số thực a, b, c.
- Đầu ra: 0, 1, hoặc 2. Nếu phương trình suy biến thành 0 = 0 (tức a == 0, b == 0, c == 0), trả về float('inf').
- Quy tắc:
- Nếu a != 0, dùng biệt thức D = b^2 - 4ac.
- Nếu a == 0 và b != 0, đó là phương trình bậc nhất với đúng một nghiệm thực.
- Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 

```python
def quadratic_roots_count(a, b, c):
    # Trường hợp suy biến
    if a == 0:
        if b == 0:
            if c == 0:
                return float('inf')  # 0 = 0 → vô số nghiệm
            else:
                return 0             # vô nghiệm
        else:
            return 1                 # bậc nhất có 1 nghiệm

    # Bậc hai
    d = b*b - 4*a*c

    if d > 0:
        return 2
    elif d == 0:
        return 1
    else:
        return 0
```

Bài 30:Yêu cầu: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu
chấm động, ngược lại trả về 'neither'.
- Đầu vào: chuỗi s.
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}.
- Quy tắc:
- Cho phép có dấu + hoặc - ở đầu.
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100').
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14').
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này. 

```python
def detect_number_type(s):
    if len(s) == 0:
        return "neither"
    
    # xử lý dấu + -
    if s[0] in "+-":
        s = s[1:]
    
    if len(s) == 0:
        return "neither"
    
    # int
    if s.isdigit():
        return "int"
    
    # float
    if s.count('.') == 1:
        left, right = s.split('.')
        if left.isdigit() and right.isdigit() and len(left) > 0 and len(right) > 0:
            return "float"
    
    return "neither"
```

Bài 32:Yêu cầu: Viết hàm is_right_triangle(a, b, c) trả về True nếu a, b, c có thể là ba cạnh của một tam giác vuông, ngược lại trả
về False.
- Đầu vào: các số dương a, b, c.
- Đầu ra: giá trị boolean.
- Lưu ý: Kiểm tra bất đẳng thức tam giác và định lý Pythagore. Dùng sai số nhỏ khi so sánh số thực. 

```python
def is_right_triangle(a, b, c):
    # sắp xếp để c là cạnh lớn nhất
    x, y, z = sorted([a, b, c])
    
    # kiểm tra điều kiện tam giác
    if x + y <= z:
        return False
    
    # kiểm tra Pythagore (có sai số nhỏ)
    eps = 1e-9
    return abs(x*x + y*y - z*z) < eps
```

Bài 33:Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op.
- Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}.
- Đầu ra: kết quả số học.
- Yêu cầu thêm:
- Nếu op là '/' và b == 0, phát sinh ValueError.
- Nếu op không hợp lệ, phát sinh ValueError. 

```python
def compute(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            raise ValueError("Division by zero")
        return a / b
    else:
        raise ValueError("Invalid operator")
```

Bài 35:Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao
cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %).
- Đầu vào: số nguyên dividend, divisor (divisor khác 0).
- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2).

```python
def divide_with_remainder(dividend, divisor):
    if divisor == 0:
        raise ValueError("Divisor cannot be zero")
    
    q = dividend // divisor
    r = dividend % divisor
    
    return (q, r)
```

Bài 41:Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái.
Giữ nguyên chữ hoa/chữ thường và quay vòng trong phạm vi A..Z hoặc a..z.

```python
def shift_char(x, n):
    if not isinstance(x, str) or len(x) != 1 or not x.isalpha():
        raise ValueError("Invalid input")
    if x.islower():
        base = ord('a')
    else:
        base = ord('A')
    
    return chr((ord(x) - base + n) % 26 + base)
```

Bài 43:Viết hàm tính diện tích tam giác có ba cạnh a, b, c (dùng công thức Heron). Phát sinh ValueError nếu ba cạnh không thể
tạo thành tam giác. 

```python
import math

def triangle_area(a, b, c):
    # kiểm tra tam giác
    if a <= 0 or b <= 0 or c <= 0:
        raise ValueError("Invalid triangle sides")
    
    if a + b <= c or a + c <= b or b + c <= a:
        raise ValueError("Not a triangle")
    
    # công thức Heron
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    
    return area
```

Bài 44:Viết hàm trả về tích các số nguyên từ 1 đến n (giai thừa). Với n=0, trả về 1. 


```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    
    res = 1
    for i in range(1, n + 1):
        res *= i
    
    return res
```

Bài 45:Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4)

```python
def count_k(n):
    if n <= 1:
        return 0
    
    return (n - 1) - (n - 1) // 2
```
