W8A16:Cho một mảng số nguyên không âm. Gọi
mảng đó là "mảng đẹp" nếu mỗi phần tử
x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp
hay không. Nếu không hãy xóa đi y số (y
là nhỏ nhất có thể) để mảng còn lại thành
mảng đẹp.


```python
from collections import Counter

def min_delete_to_beautiful(arr):
    freq = Counter(arr)
    delete = 0

    for x, f in freq.items():
        if f >= x:
            delete += f - x
        else:
            delete += f

    return delete


# Ví dụ
arr = [3, 3, 3, 2, 2, 1]
print(min_delete_to_beautiful(arr))
```

W8A17:Tìm phần tử xuất hiện nhiều hơn n/2 lần
trong mảng (nếu có) với độ phức tạp thời
gian là O(n), độ phức tạp bộ nhớ là O(1),
- Boyer–Moore. Nếu không có phần tử
nào thỏa mãn thì không in gì

```python
n = int(input())
a = list(map(int, input().split()))

candidate = None
count = 0

for x in a:
    if count == 0:
        candidate = x
        count = 1
    elif x == candidate:
        count += 1
    else:
        count -= 1

if a.count(candidate) > n // 2:
    print(candidate)
```

W8A18:Sắp xếp có điều kiện: Số chẵn đứng
trước, số lẻ đứng sau, chẵn tăng dần, lẻ
giảm dần

```python
n = int(input())
a = list(map(int, input().split()))

chan = []
le = []

for x in a:
    if x % 2 == 0:
        chan.append(x)
    else:
        le.append(x)

chan.sort()
le.sort(reverse=True)

kq = chan + le

for x in kq:
    print(x, end=' ')
```

W8A19:Merge 2 mảng đã sắp xếp (có thể tăng
dần, có thể giảm dần) mà không dùng
bất cứ thuật toán sắp xếp nào thành một
mảng cũng có cùng thứ tự sắp xếp như 2
mảng ban đầu

```python
n = int(input())
a = list(map(int, input().split()))

m = int(input())
b = list(map(int, input().split()))

kq = []

i = 0
j = 0

tang = True
if n > 1 and a[0] > a[1]:
    tang = False

while i < n and j < m:
    if tang:
        if a[i] <= b[j]:
            kq.append(a[i])
            i += 1
        else:
            kq.append(b[j])
            j += 1
    else:
        if a[i] >= b[j]:
            kq.append(a[i])
            i += 1
        else:
            kq.append(b[j])
            j += 1

while i < n:
    kq.append(a[i])
    i += 1

while j < m:
    kq.append(b[j])
    j += 1

print(*kq)
```

W8A22:Cho N tên sinh viên gồm cả họ và tên,
tìm xem tên nào hay gặp nhất. Nếu có
nhiều tên cùng số lần gặp thì lấy tên có
thứ tự từ điển nhỏ nhất

```python
n = int(input())

d = {}

for _ in range(n):
    ho_ten = input().strip()
    ten = ho_ten.split()[-1]

    if ten in d:
        d[ten] += 1
    else:
        d[ten] = 1

max_dem = max(d.values())

kq = ""

for ten in d:
    if d[ten] == max_dem:
        if kq == "" or ten < kq:
            kq = ten

print(kq)
```

W8A25:Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số


```python
s = input().strip()
k = int(input())

xoa = len(s) - k
st = []

for c in s:
    while st and xoa > 0 and st[-1] < c:
        st.pop()
        xoa -= 1
    st.append(c)

while xoa > 0:
    st.pop()
    xoa -= 1

print(''.join(st[:k]))
```

W8A27:Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra
xem dãy ngoặc đã cho có hợp lệ hay
không


```python
s = input()

st = []
ok = True

for c in s:
    if c in "({[":
        st.append(c)
    else:
        if len(st) == 0:
            ok = False
            break

        top = st.pop()

        if (c == ')' and top != '(') or \
           (c == '}' and top != '{') or \
           (c == ']' and top != '['):
            ok = False
            break

if len(st) > 0:
    ok = False

print("YES" if ok else "NO")
```

W8A28:Bà
̀i toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
def hop_so(n):
    if n <= 3:
        return False

    i = 2
    while i * i <= n:
        if n % i == 0:
            return True
        i += 1

    return False


N = int(input())

max_len = 0
max_start = 0

cur_len = 0
cur_start = 0

for i in range(1, N + 1):
    if hop_so(i):
        if cur_len == 0:
            cur_start = i
        cur_len += 1

        if cur_len > max_len:
            max_len = cur_len
            max_start = cur_start
    else:
        cur_len = 0

for i in range(max_start, max_start + max_len):
    if i > max_start:
        print(",", end=" ")
    print(i, end="")
```

W8A29:Tìm Dãy con tăng dần liên tục dài nhất
của một danh sách nhập vào từ bàn
phím. Lấy dãy đầu tiên nếu có nhiều dãy
có số phần tử bằng nhau

```python
a = list(map(int, input().split()))

max_start = 0
max_len = 1

cur_start = 0
cur_len = 1

for i in range(1, len(a)):
    if a[i] > a[i - 1]:
        cur_len += 1
    else:
        if cur_len > max_len:
            max_len = cur_len
            max_start = cur_start

        cur_start = i
        cur_len = 1

if cur_len > max_len:
    max_len = cur_len
    max_start = cur_start

print(*a[max_start:max_start + max_len])
```

W8A30:Tìm Dãy con tăng dần KHÔNG liên tục
dài nhất của một danh sách nhập vào từ
bàn phím

```python
a = list(map(int, input().split()))
n = len(a)

dp = [1] * n
truoc = [-1] * n

for i in range(n):
    for j in range(i):
        if a[j] < a[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            truoc[i] = j

vt = 0
for i in range(1, n):
    if dp[i] > dp[vt]:
        vt = i

kq = []

while vt != -1:
    kq.append(a[vt])
    vt = truoc[vt]

kq.reverse()

print(*kq)
```

W8A31:Đổi chỗ đúng 2 chữ số để được số nhỏ
nhất có thể


```python
s = list(input().strip())
n = len(s)

done = False

for i in range(n):
    mn = s[i]
    pos = i

    for j in range(i + 1, n):
        if s[j] < mn:
            if i == 0 and s[j] == '0':
                continue
            mn = s[j]
            pos = j
        elif s[j] == mn:
            pos = j

    if pos != i:
        s[i], s[pos] = s[pos], s[i]
        done = True
        break

print(''.join(s))
```
