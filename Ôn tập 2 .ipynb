Trả về dưới dạng danh sách [max_value, min_value] (không dùng tuple).
Với số nhỏ nhất, không được để chữ số 0 đứng đầu (ví dụ: 2015 → 5210 và 1025)

```python
def sap_xep_so(n):
    digits = list(str(n))

    # Số lớn nhất
    max_value = int(''.join(sorted(digits, reverse=True)))

    # Số nhỏ nhất
    digits.sort()
    if digits[0] == '0':
        for i in range(1, len(digits)):
            if digits[i] != '0':
                digits[0], digits[i] = digits[i], digits[0]
                break

    min_value = int(''.join(digits))

    return [max_value, min_value]
n = int(input())
print(sap_xep_so(n))
```

Bài 8:Yêu cầu: Cho chuỗi thời gian dạng "hh:mm:ss", hãy trả về tổng số giây.
Giờ, phút, giây phải là số nguyên với 0 ≤ mm < 60 và 0 ≤ ss < 60.
Trả về -1 nếu giá trị nằm ngoài phạm vi hợp lệ.


```python
def doi_ra_giay(t):
    parts = t.split(":")

    if len(parts) != 3:
        return -1

    h = int(parts[0])
    m = int(parts[1])
    s = int(parts[2])

    if m < 0 or m >= 60 or s < 0 or s >= 60 or h < 0:
        return -1

    return h * 3600 + m * 60 + s
```

- Dùng vòng lặp for để so sánh các vị trí đối xứng.
- Không dùng tuple hoặc dictionary

```python
def is_palindrome(a):
    n = len(a)

    for i in range(n // 2):
        if a[i] != a[n - 1 - i]:
            return False

    return True
```

```python
def rotate_right(a, x):
    n = len(a)

    if n == 0:
        return []

    x = x % n

    return a[-x:] + a[:-x]
```

Bài 20:Bạn được cho hai danh sách:
- A: người dùng đã đăng ký
- B: người dùng đã thanh toán
Hãy trả về 3 danh sách:
1. Người đã thanh toán nhưng chưa đăng ký
2. Người đã đăng ký nhưng chưa thanh toán
3. Người vừa đăng ký và đã thanh toán
Mỗi danh sách kết quả cần được sắp xếp theo thứ tự tăng dần

```python
def classify_users(A, B):
    setA = set(A)  # đã đăng ký
    setB = set(B)  # đã thanh toán

    pay_not_reg = sorted(setB - setA)   # thanh toán nhưng chưa đăng ký
    reg_not_pay = sorted(setA - setB)   # đăng ký nhưng chưa thanh toán
    both = sorted(setA & setB)          # vừa đăng ký vừa thanh toán

    return pay_not_reg, reg_not_pay, both
```

Bài 25:Viết hàm is_prime(n) trả về True nếu n là số nguyên tố và False nếu ngược lại. Theo quy ước, các số nhỏ hơn 2 không phải là số nguyên
tố.


```python
def is_prime(n):
    if n < 2:
        return False

    if n == 2:
        return True

    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True
```

```python
def is_perfect(n):
    if n <= 0:
        return False

    tong = 0
    for i in range(1, n):
        if n % i == 0:
            tong += i

    return tong == n
```

Giả sử các thành phần được ngăn cách bởi dấu cách; có thể có nhiều dấu cách liên tiếp. Nếu chỉ có một thành phần thì giữ nguyên.

```python
def reorder_name(s):
    s = str(s).strip()

    parts = s.split()

    if len(parts) <= 1:
        return s

    ho = parts[0]
    ten = parts[-1]
    dem = parts[1:-1]

    return " ".join([ten] + dem + [ho])
```

Bài 30:Yêu cầu: Cài đặt giải mã Atbash cho các chữ cái tiếng Anh. Ánh xạ a→z, b→y, ..., z→a và tương tự A→Z, ..., Z→A.
Các ký tự không phải chữ cái được giữ nguyên

```python
def atbash_decode(s):
    result = []

    for ch in str(s):
        if 'a' <= ch <= 'z':
            result.append(chr(ord('z') - (ord(ch) - ord('a'))))
        elif 'A' <= ch <= 'Z':
            result.append(chr(ord('Z') - (ord(ch) - ord('A'))))
        else:
            result.append(ch)

    return "".join(result)
```

Bài 31:Yêu cầu: Một chuỗi được gọi là perfect nếu nó chứa ít nhất một chữ hoa, ít nhất một chữ thường và ít nhất một chữ số.
Trả về True nếu chuỗi là perfect; ngược lại trả về False. Nếu đầu vào không phải chuỗi, hãy chuyển thành chuỗi.

```python
def is_perfect_string(s):
    s = str(s)

    has_upper = False
    has_lower = False
    has_digit = False

    for ch in s:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit = True

    return has_upper and has_lower and has_digit
```

Bài 32:Liệt kê tất cả số nguyên tố nhỏ hơn x (x < 1000).
Viết hàm primes_less_than(x) trả về danh sách tất cả các số nguyên tố nhỏ hơn x.
- Dùng các vòng lặp đơn giản và phép kiểm tra số nguyên.
- Không dùng tuple hoặc dictionary

```python
def primes_less_than(x):
    if x <= 2:
        return []

    result = []

    for n in range(2, x):
        is_prime = True

        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                is_prime = False
                break

        if is_prime:
            result.append(n)

    return result
```

Bài 34:Bạn được cho mảng a gồm n số nguyên.
Trong một bước, bạn có thể chọn i != j rồi gán a[i] := a[j].
Bạn có thể thực hiện thao tác này bao nhiêu lần cũng được.
Yêu cầu: Trả về True nếu có thể biến đổi mảng để tổng các phần tử là số lẻ.
Khai báo hàm: can_make_sum_odd(a) -> bool

```python
def can_make_sum_odd(a):
    has_even = False
    has_odd = False

    for x in a:
        if x % 2 == 0:
            has_even = True
        else:
            has_odd = True

    # Nếu có cả chẵn và lẻ → luôn có thể tạo tổng lẻ
    if has_even and has_odd:
        return True

    # Nếu chỉ toàn lẻ: tổng lẻ khi n lẻ
    if has_odd:
        return len(a) % 2 == 1

    # Nếu toàn chẵn: không thể tạo tổng lẻ
    return False
```

Bài 35:Bạn được cho số nguyên dương chẵn n.
Hãy xây dựng mảng a có độ dài n sao cho:
- n/2 phần tử đầu là số chẵn
- n/2 phần tử cuối là số lẻ
- tất cả các phần tử đều khác nhau và dương
- tổng nửa đầu bằng tổng nửa sau
Nếu không thể, trả về (False, []). Nếu có thể, trả về (True, a).

```python
def build_array(n):
    if n % 4 != 0:
        return (False, [])

    a = []

    half = n // 2

    # nửa đầu: số chẵn
    first = []
    for i in range(1, half + 1):
        first.append(2 * i)

    # nửa sau: số lẻ
    second = []
    for i in range(1, half):
        second.append(2 * i - 1)

    # phần tử cuối điều chỉnh để cân bằng tổng
    sum_first = sum(first)
    sum_second = sum(second)

    second.append(sum_first - sum_second)

    a = first + second

    return (True, a)
```

Bài 38:Viết hàm inverted_triangle(n) trả về một tam giác ngược cân bằng gồm các dấu sao *, có độ rộng là n.
- n phải là một số nguyên dương lẻ, nếu không thì trả về chuỗi rỗng.
- Mỗi dòng có đúng n ký tự, số dấu sao giảm dần 2 đơn vị mỗi dòng cho đến khi còn 1 dấu sao.


```python
def inverted_triangle(n):
    if n <= 0 or n % 2 == 0:
        return ""

    lines = []

    for stars in range(n, 0, -2):
        line = "*" * stars
        lines.append(line)

    return "\n".join(lines)
```

```python
def sum_two_diagonals(mat):
    n = len(mat)
    tong = 0

    for i in range(n):
        tong += mat[i][i]              # đường chéo chính
        tong += mat[i][n - 1 - i]     # đường chéo phụ

    # nếu n lẻ thì ô giữa bị cộng 2 lần → trừ đi 1 lần
    if n % 2 == 1:
        tong -= mat[n // 2][n // 2]

    return tong
```

Bài 40:Tổng các phần tử a[i][j] mà i + j là số chẵn.
Viết hàm sum_where_i_plus_j_even(mat) trả về tổng các phần tử ở những vị trí
mà i + j là số chẵn (dùng chỉ số bắt đầu từ 0; tính chẵn lẻ cũng tương ứng với cách đánh số từ 1).

- Không dùng tuple hoặc dictionary.

```python
def sum_where_i_plus_j_even(mat):
    n = len(mat)
    tong = 0

    for i in range(n):
        for j in range(len(mat[i])):
            if (i + j) % 2 == 0:
                tong += mat[i][j]

    return tong
```

Bài 41:Tổng các phần tử nằm trên hàng lẻ hoặc cột lẻ (đánh số từ 1).
Viết hàm sum_odd_rows_or_cols(mat) trả về tổng các phần tử mà chỉ số hàng hoặc
chỉ số cột là số lẻ nếu dùng cách đánh số bắt đầu từ 1. (Tức là lấy mọi ô mà hàng lẻ hoặc cột lẻ.)
- Dùng vòng lặp.
- Không dùng tuple hoặc dictionary

```python
def sum_odd_rows_or_cols(mat):
    n = len(mat)
    m = len(mat[0]) if n > 0 else 0

    tong = 0

    for i in range(n):
        for j in range(m):
            if i % 2 == 0 or j % 2 == 0:
                tong += mat[i][j]

    return tong
```

Bài 42:Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m.
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất.
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất.
- Dùng vòng lặp để tính tổng từng cột.
- Không dùng tuple hoặc dictionary.

```python
def column_with_max_sum(mat):
    n = len(mat)
    if n == 0:
        return -1

    m = len(mat[0])

    max_sum = None
    best_col = 0

    for j in range(m):
        tong = 0
        for i in range(n):
            tong += mat[i][j]

        if max_sum is None or tong > max_sum:
            max_sum = tong
            best_col = j

    return best_col
```

Bài 43:Viết hàm zigzag_numbers(n) trả về một chuỗi biểu diễn ma trận n x n chứa các số từ 1 đến n^2. Các hàng cách nhau bởi ký tự xuống
dòng. Với chỉ số hàng bắt đầu từ 0: hàng chẵn đi từ trái sang phải, hàng lẻ đi từ phải sang trái. Các số trong cùng một hàng cách nhau bởi
một dấu cách.


```python
def zigzag_numbers(n):
    if n <= 0:
        return ""

    result = []
    num = 1

    for i in range(n):
        row = []

        for j in range(n):
            row.append(str(num))
            num += 1

        if i % 2 == 1:
            row.reverse()

        result.append(" ".join(row))

    return "\n".join(result)
```

Bài 44:Viết hàm count_solutions_3vars(a, b, c, t) trả về số nghiệm nguyên không âm (x, y, z) của phương trình a*x + b*y + c*z = t. Giả sử a, b, c,
t là các số nguyên. Nếu a <= 0 hoặc b <= 0 hoặc c <= 0 hoặc t < 0 thì trả về 0.


```python
def count_solutions_3vars(a, b, c, t):
    if a <= 0 or b <= 0 or c <= 0 or t < 0:
        return 0

    count = 0

    for x in range(t // a + 1):
        for y in range(t // b + 1):
            remaining = t - a * x - b * y

            if remaining < 0:
                break

            if remaining % c == 0:
                z = remaining // c
                if z >= 0:
                    count += 1

    return count
```

Bài 45:Số lần đổi chỗ kề nhau ít nhất để đưa số 1 duy nhất vào giữa ma trận có kích thước lẻ.
Bạn được cho ma trận n x n (với n là số lẻ). Ma trận chứa đúng một ô có giá trị 1,
còn tất cả các ô khác đều là 0. Trong một thao tác, bạn có thể đổi chỗ hai hàng kề nhau hoặc hai cột kề nhau.
Hãy tính số thao tác ít nhất để đưa số 1 vào đúng ô trung tâm

```python
def min_swaps_to_center(mat):
    n = len(mat)

    # tìm vị trí số 1
    for i in range(n):
        for j in range(n):
            if mat[i][j] == 1:
                x, y = i, j
                break

    center = n // 2

    return abs(x - center) + abs(y - center)
```

Bài 46:Yêu cầu: Cho danh sách tên sinh viên, hãy trả về 3 tên xuất hiện nhiều nhất.
Nếu có ít hơn 3 tên khác nhau, hãy trả về tất cả.
Yêu cầu thêm: Ưu tiên dùng map/dictionary.
Khai báo hàm: top3_frequent_names(names) -> list[str]


```python
def top3_frequent_names(names):
    freq = {}

    for name in names:
        if name in freq:
            freq[name] += 1
        else:
            freq[name] = 1

    sorted_names = sorted(freq.items(), key=lambda x: x[1], reverse=True)

    result = []
    for i in range(min(3, len(sorted_names))):
        result.append(sorted_names[i][0])

    return result
```

Bài 47:Yêu cầu: Cho danh sách các bộ ba (tuple gồm 3 số nguyên), hãy trả về bộ ba xuất hiện nhiều nhất.
Khai báo hàm: most_frequent_triple(triples) -> tuple[int,int,int]
- Phát sinh ValueError nếu danh sách đầu vào rỗng.

```python
def most_frequent_triple(triples):
    if len(triples) == 0:
        raise ValueError("Empty list")

    freq = {}

    for t in triples:
        if t in freq:
            freq[t] += 1
        else:
            freq[t] = 1

    most_common = None
    max_count = 0

    for t in freq:
        if freq[t] > max_count:
            max_count = freq[t]
            most_common = t

    return most_common
```

Bài 48:Cho danh sách số nguyên nums và số nguyên k, hãy trả về True nếu tồn tại hai chỉ số khác nhau i và j sao cho:
- nums[i] == nums[j], và
- abs(i - j) <= k
Khai báo hàm: contains_nearby_duplicate(nums, k) -> bool

```python
def contains_nearby_duplicate(nums, k):
    last_index = {}

    for i, value in enumerate(nums):
        if value in last_index and i - last_index[value] <= k:
            return True

        last_index[value] = i

    return False
```

Bài 49:Bạn được cho hai mảng số nguyên 2 chiều items1 và items2.
Mỗi phần tử có dạng [value, weight], và value là duy nhất trong từng danh sách.
Yêu cầu: Trả về danh sách 2 chiều ret, trong đó mỗi phần tử là [value, total_weight],
với total_weight là tổng trọng lượng của cùng value trong cả hai danh sách.
Kết quả cần được sắp theo value tăng dần.

```python
def merge_items(items1, items2):
    weight_map = {}

    # cộng từ items1
    for value, weight in items1:
        if value in weight_map:
            weight_map[value] += weight
        else:
            weight_map[value] = weight

    # cộng từ items2
    for value, weight in items2:
        if value in weight_map:
            weight_map[value] += weight
        else:
            weight_map[value] = weight

    # tạo kết quả và sắp xếp theo value tăng dần
    result = []
    for value in sorted(weight_map):
        result.append([value, weight_map[value]])

    return result
```
