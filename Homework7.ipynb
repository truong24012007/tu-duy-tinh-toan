W7A16:Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
from collections import Counter

def min_delete_beautiful(arr):
    freq = Counter(arr)
    deleted = 0

    for x, cnt in freq.items():
        if cnt < x:
            deleted += cnt
        elif cnt > x:
            deleted += cnt - x

    return deleted
```

W7A17:Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì

```python
def majority_element(arr):
    candidate = None
    count = 0

    # Tìm ứng viên
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1

    # Kiểm tra ứng viên
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)
```

W7A18:
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ;

```python
def conditional_sort(arr):
    even = []
    odd = []

    for x in arr:
        if x % 2 == 0:
            even.append(x)
        else:
            odd.append(x)

    even.sort()          # tăng dần
    odd.sort(reverse=True)  # giảm dần

    return even + odd
```

W7A19:
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào


```python
def merge_ascending(a, b):
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    while i < len(a):
        result.append(a[i])
        i += 1

    while j < len(b):
        result.append(b[j])
        j += 1

    return result
```

W7A22:
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())

count = {}

for _ in range(n):
    fullname = input().strip()
    name = fullname.split()[-1]  # lấy tên

    count[name] = count.get(name, 0) + 1

max_freq = max(count.values())

result = min(name for name, freq in count.items() if freq == max_freq)

print(result)
```

W7A25:Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
def largest_k_digits(s, k):
    remove = len(s) - k
    stack = []

    for digit in s:
        while remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            remove -= 1

        stack.append(digit)

    return ''.join(stack[:k])
```

W7A27:Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
def is_valid(s):
    stack = []
    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for ch in s:
        if ch in '([{':
            stack.append(ch)
        else:
            if not stack or stack[-1] != pairs[ch]:
                return False
            stack.pop()

    return len(stack) == 0


s = input()
print(is_valid(s))
```

W7A28:Bà
̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
def is_composite(x):
    if x < 4:
        return False

    i = 2
    while i * i <= x:
        if x % i == 0:
            return True
        i += 1

    return False


N = int(input())

best_start = -1
best_len = 0

cur_start = -1
cur_len = 0

for x in range(1, N + 1):
    if is_composite(x):
        if cur_len == 0:
            cur_start = x
        cur_len += 1
    else:
        if cur_len > best_len:
            best_len = cur_len
            best_start = cur_start
        cur_len = 0

# xử lý trường hợp dãy dài nhất nằm ở cuối
if cur_len > best_len:
    best_len = cur_len
    best_start = cur_start

if best_len > 0:
    for x in range(best_start, best_start + best_len):
        print(x, end=" ")
```

W7A29:Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
arr = list(map(int, input().split()))

best_start = 0
best_len = 1

cur_start = 0
cur_len = 1

for i in range(1, len(arr)):
    if arr[i] > arr[i - 1]:
        cur_len += 1
    else:
        if cur_len > best_len:
            best_len = cur_len
            best_start = cur_start

        cur_start = i
        cur_len = 1

if cur_len > best_len:
    best_len = cur_len
    best_start = cur_start

print(*arr[best_start:best_start + best_len])
```

W7A30:Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
arr = list(map(int, input().split()))
n = len(arr)

dp = [1] * n
parent = [-1] * n

for i in range(n):
    for j in range(i):
        if arr[j] < arr[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            parent[i] = j

# tìm vị trí kết thúc LIS
end = 0
for i in range(1, n):
    if dp[i] > dp[end]:
        end = i

# truy vết dãy
lis = []
while end != -1:
    lis.append(arr[end])
    end = parent[end]

lis.reverse()

print(*lis)
```
